Écrire une fonction qui reçoit un nombre `x` et retourne :
- `'Positif'` si `x > 0`
- `'Négatif'` si `x < 0`
- `'Zéro'` sinon



```python
def fonction_positif_negatif_zero():
    x=input("Introduire la valeur de x :")
    x=float(x)
    if x>0:
        print("x est Positif")
    elif x<0:
        print("x est Negatif")
    else:
        print("x est zero")
fonction_positif_negatif_zero()
```

## Exercice 12 (2 pts)

Écrire une fonction qui reçoit une liste de nombres saisis successivement et retourne la somme jusqu’à rencontrer un 0.

Exemple : `[5, 2, -1, 0, 100]` -> somme = `6`.

Consigne : utiliser une boucle `while`.




```python
def somme_jusqua_zero(liste):
    somme = 0
    i = 0
    while i < len(liste) and liste[i] != 0:
        somme += liste[i]
        i += 1
    return somme
exemple = [10, 23, 1, 3, -7,0,22,53,22,66]
print(f" la somme jusqu'à 0 c'est : {somme_jusqua_zero(exemple)}")
```
